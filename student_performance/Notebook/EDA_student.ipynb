```python

# Find relation between race_ethnicity and Avg score
avg_score_by_race_ethnicity = df.groupby('race_ethnicity')['Avg'].mean()

ax = avg_score_by_race_ethnicity.plot(kind='bar',figsize=(10, 6))

# Show values on top of bars
ax.bar_label(ax.containers[0], fmt='%.2f', padding=3)

plt.title('Avg Score by Race_Ethnicity')
plt.xlabel('Race_Ethnicity')
plt.ylabel('Avg Score')

plt.show()
```

This analysis compares the average academic performance of students across different race_ethnicity groups.
and group E perform better as comapre to other groups

```python
#pass rate by lunch type
pass_rate_lunch = df.groupby('lunch')['passed'].mean() * 100

ax = pass_rate_lunch.plot(kind='bar', figsize=(8,5))

ax.bar_label(ax.containers[0], fmt='%.2f%%', padding=3)

plt.title('Pass Rate by Lunch Type')
plt.xlabel('Lunch Type')
plt.ylabel('Pass Rate (%)')

plt.show()
```
